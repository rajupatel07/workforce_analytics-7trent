```python
from pyspark.sql import functions as F

people = spark.table(
    "workforce_analytics.silver.dim_people"
)

dim_people_gold = people.select(
    F.col("people_id"),
    F.col("employee_name"),
    F.col("first_name"),
    F.col("surname"),
    F.col("job_role"),
    F.col("employment_type"),
    F.col("team"),
    F.col("hire_date")
)

(
    dim_people_gold.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable("workforce_analytics.gold.dim_people_gold")
)

print("dim_people_gold created successfully!")
```

```python
gold_people = spark.table(
    "workforce_analytics.gold.dim_people_gold"
)

print("Gold people rows:", gold_people.count())

display(gold_people.limit(10))
```
