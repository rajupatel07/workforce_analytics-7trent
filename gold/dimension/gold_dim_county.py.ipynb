```python
from pyspark.sql import functions as F

county = spark.table(
    "workforce_analytics.silver.dim_county"
)

dim_county_gold = county.select(
    F.col("county_id"),
    F.col("county_name"),
    F.col("region"),
    F.col("country"),
    F.col("population_band"),
    F.col("urban_rural")
)

(
    dim_county_gold.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable("workforce_analytics.gold.dim_county_gold")
)

print("dim_county_gold created successfully!")
```
