```python
from pyspark.sql import functions as F

shrinkage_type = spark.table(
    "workforce_analytics.silver.dim_shrinkage_type"
)

dim_shrinkage_type_gold = shrinkage_type.select(
    F.col("shrinkage_type_id"),
    F.col("shrinkage_type"),
    F.col("category"),
    F.col("is_paid"),
    F.col("is_planned")
)

(
    dim_shrinkage_type_gold.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable("workforce_analytics.gold.dim_shrinkage_type_gold")
)

print("dim_shrinkage_type_gold created successfully!")
```
